```python
import pandas as pd
df=pd.read_csv("../data/clean_reviews.csv")
df.head()
```

```python
df["review"].head(10).tolist()
```

```python
import re
def light_clean(text):
    text =str(text)
    text = re.sub(r"<.*?>"," ",text)
    text = re.sub(r"https?://\S+|www\.\S+"," ",text)
    text = re.sub(r"\s+"," ",text).strip()
    return text
```

```python
light_clean(df["review"].iloc[0])
```

```python
df["clean_review"] = df["review"].apply(light_clean)
```
