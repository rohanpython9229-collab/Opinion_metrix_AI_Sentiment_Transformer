```python
from src.inference import predict_sentiment
```

```python
test_reviews = [
    "This product is amazing and works perfectly!",
    "The product is okay, nothing special.",
    "Terrible quality, completely disappointed with this purchase."
]

for review in test_reviews:
    result = predict_sentiment(review)

    print("Review:", review)
    print("Sentiment:", result["sentiment"])
    print("Confidence:", round(result["confidence"], 4))
    print("-" * 60)
```

```python
custom_reviews = [
    "The battery lasts all day and the sound quality is excellent.",
    "The product arrived on time, but the quality is just average.",
    "I regret buying this. The device stopped working after two days.",
    "Good value for money, although the packaging could be better.",
    "Absolutely loved it! The build quality is fantastic.",
    "It works, but there is nothing particularly impressive about it.",
    "Worst purchase ever. Completely useless and poorly made.",
    "The headphones are comfortable, but the microphone quality is poor.",
    "Fast delivery and excellent product. Very happy with my purchase.",
    "The product is decent for the price, but I expected more."
]
```

```python
for i, review in enumerate(custom_reviews, start=1):
    result = predict_sentiment(review)

    print(f"{i}. Review: {review}")
    print(f"   Sentiment: {result['sentiment']}")
    print(f"   Confidence: {result['confidence']:.2%}")
    print("-" * 70)
```
